```python
# Importer librairies standards
import pandas as pd
import numpy as np
import math
import bs4 as bs
import urllib.request
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from ipywidgets import interact, interact_manual
```

```python
# Téléchargement des datasets
interactions_train = pd.read_csv('https://raw.githubusercontent.com/Vane-s/UNIL_Rolex/main/interactions_train.csv')
items = pd.read_csv("https://raw.githubusercontent.com/Vane-s/UNIL_Rolex/main/items.csv")
# Les premières lignes de chaque dataset
display(interactions_train.head())
display(items.head())
```

```python
# User-item interaction matrix
user_item_matrix = interactions_train.groupby(['u', 'i'])['t'].sum().unstack().fillna(0)

# Générer des recommandations
def get_top_n_recommendations(user_prediction, n=10):
    recommendations = []
    for user_id in user_prediction.index:
        user_scores = user_prediction.loc[user_id]
        top_items = user_scores.sort_values(ascending=False).head(n).index.tolist()
        recommendations.append({'user_id': user_id, 'recommendation': ' '.join(map(str, top_items))})
    return recommendations

# Le top 10 des recommendations
top_recommendations = get_top_n_recommendations(user_item_matrix, n=10)

# Dataframe pour les recommandations finales
final_recommendations = pd.DataFrame(top_recommendations)

# Recommandation en CSV
final_recommendations.to_csv("submission.csv", index=False, sep=",", quotechar="", quoting=3)
print("Fichier CSV 'submission.csv' généré avec succès.")
```
